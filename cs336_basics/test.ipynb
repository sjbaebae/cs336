```python
import importlib
import tokenizer
importlib.reload(tokenizer)
# reload Tokenizer
from tokenizer import Tokenizer
import pandas as pd
import json
import regex
import pathlib
import os
```

```python
tiny_path = "../outputs/tinystories"
tiny_vocab_path = os.path.join(tiny_path, "vocab.pkl")
tiny_merges_path = os.path.join(tiny_path, "merges.pkl")

owt_path = "../outputs/openwebtext"
owt_vocab_path = os.path.join(owt_path, "vocab.pkl")
owt_merges_path = os.path.join(owt_path, "merges.pkl")

tiny_tokenizer = Tokenizer.from_file(tiny_vocab_path, tiny_merges_path)
owt_tokenizer = Tokenizer.from_file(owt_vocab_path, owt_merges_path)
```

```python
# load in the sample data
tiny_samples = pd.read_csv("../samples/tinystories_10_examples.csv")
owt_samples = pd.read_csv("../samples/owt_samples.csv")
```

```python
tiny_sample_texts = "".join(tiny_samples["text"]) # tokenize into new sample_csv
owt_sample_texts = "".join(owt_samples["text"])

# compress with tokenization
tiny_sample_texts_encoded = tiny_tokenizer.encode(tiny_sample_texts)
owt_sample_texts_encoded = owt_tokenizer.encode(owt_sample_texts)
```

```python
import time

t0 = time.perf_counter()
tiny_sample_texts_encoded = tiny_tokenizer.encode(tiny_sample_texts)

time_taken = time.perf_counter() - t0
print("time taken: ", time_taken)
tiny_sample_bytes = len(tiny_sample_texts.encode("utf-8"))

print(f"bytes (MB) / s encoding: {(tiny_sample_bytes / time_taken / (1024 ** 2)):.2f}")
```

```python
len_tiny_bytes = len(tiny_sample_texts.encode("utf-8"))
len_owt_bytes = len(owt_sample_texts.encode("utf-8"))

len_tiny_ids_u16 = len(tiny_sample_texts_encoded) * 2
len_owt_ids_u16 = len(owt_sample_texts_encoded) * 2

print(f"Tinystories Compression Ratio 10 Samples: {(len_tiny_ids_u16/ len_tiny_bytes * 100):.2f}%")
print(f"OpenWebText Compression Ratio 10 Samples: {(len_owt_ids_u16 / len_owt_bytes * 100):.2f}%")
```

```python
# compress tinystories with OWT tokenizer
tiny_sample_with_owttokenizer_texts_encoded = owt_tokenizer.encode(tiny_sample_texts)
len_tiny_owt_ids_u16 = len(tiny_sample_with_owttokenizer_texts_encoded) * 2

print(f"Tiny sample with OWT tokenizer Compression Ratio 10 Samples: {(len_tiny_owt_ids_u16 / len_tiny_bytes * 100):.2f}%")
```
